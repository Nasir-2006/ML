```python
import numpy as np
def scaling():
    miu = np.mean(X, axis=0)
    sigma = np.std(X, axis=0)
    X_norm = (X - miu) / sigma 
    return X_norm
def sigmoid_function(X, w, b):
    Z = np.dot(X, w) + b
    A = 1 / (1 + np.exp(-Z))
    return A
def regularized_cost(X, Y, w, A, b):
    m = len(X)
    lambd = 0.1
    loss = -np.sum(np.log(A) * Y + np.log(1 - A) * (1 - Y)) * (1 / m) #First part of the cost function
    #Second part of the cost function
    checker = (lambd / (2 * m)) * np.sum(np.square(w)) #Secound part of the cost function
    return loss + checker
def regulaized_gradient(X, Y, A, w, b):
    m = len(X)
    dz = A - Y
    dw = (1 / m) * np.dot(X.T, dz) #Transpose of X multiplied by dz, scaled by 1/
    db = (1 / m) * np.sum(dz)
    return dw, db
# Compute the gradients of the regularized cost function with respect to w and b
def gradient_decent(X, Y, w, b, learning_rate=0.01, num_iterations=1000):
    for i in range(num_iterations):
        A = sigmoid_function(X, w, b)
        dw, db = regulaized_gradient(X, Y, A, w, b)
        w -= learning_rate * dw
        b -= learning_rate * db
    return w, b
#Synthetic Data for Logistic Regression
X = np.array([
    [8,  2],  
    [9,  14],  
    [13, 1],  
    [14, 22], 
    [16, 6],   
    [20, 25],  
    [21, 18], 
    [23, 3]    
])

Y = np.array([0, 1, 0, 1, 0, 1, 1, 0])

X_norm = scaling()
w = np.zeros(X.shape[1])
b = 0.0
w_trained, b_trained = gradient_decent(
    X_norm, Y, w, b, learning_rate=0.1, num_iterations=2000
)
final_A = sigmoid_function(X_norm, w_trained, b_trained)
final_cost = regularized_cost(X_norm, Y, w_trained, final_A, b_trained)
predictions = (final_A >= 0.5).astype(int)
print("--- Training Results ---")
print("Learned Weights [Hour, Orders]:", np.round(w_trained, 4))
print("Learned Bias:", round(b_trained, 4))
print("Final Cost:", round(final_cost, 4))
print("Predictions: ", predictions)
print("Actual Labels:", Y)
print(f"Accuracy: {np.mean(predictions == Y) * 100:.1f}%")
```
