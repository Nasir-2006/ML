```python
import numpy as np
def sigmoid(x):
    return 1/(1+np.exp(-x))

def dense(x_input,W,b):
    units  =  W.shape[1]
    a_out = np.zeros(units)
    for i in range(units):
        w = W[:,i]
        z  = np.dot(x_input,w)+b[i]
        a_out[i] = sigmoid(z)
        return a_out
    
def sequential(x_in, W1, b1, W2, b2, W3, b3):
    a1 = dense(x_in, W1, b1)
    a2 = dense(a1, W2, b2)
    a3 = dense(a2, W3, b3)
    return a3


x_in = np.array([0.5, -1.2])
#Layer 1 ka data he
W1 = np.array([
    [0.2, -0.5, 0.8],
    [-0.1, 0.4, -0.3]
])
b1 = np.array([0.1, -0.2, 0.0])
#Layer 2 ka data he
W2 = np.array([
    [0.5, -0.3],
    [-0.7, 0.1],
    [0.9, -0.4]
])
b2 = np.array([-0.1, 0.2])
#Layer 3 ka data he
W3 = np.array([
    [1.2],
    [-0.8]
])
b3 = np.array([0.5])

output = sequential(x_in, W1, b1, W2, b2, W3, b3)

print("Final Output Probability:", output)
print("Binary Decision (>= 0.5):", int(output[0] >= 0.5))
```
