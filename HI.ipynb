```python
import numpy as np
# 1. Raw Dataset: 4 marizon ka data (m=4), 2 features (n=2: Tumor Size, Age)
X = np.array([
    [2.5, 45],
    [1.2, 22],
    [5.8, 60],
    [3.1, 35]
])

# 2. Weights (n=2): Har feature ka aik weight
w = np.array([0.5, -0.2])

# 3. Bias (Scalar)
b = -0.1
y = np.array([0, 0, 1, 1])
def scaling(X):
    mean = np.mean(X, axis=0)
    std = np.std(X, axis=0)
    X_normalized = (X - mean) / std
    return X_normalized
def sigmoid(Z):
    return 1/(1+np.exp(-Z))
def sigmoid_function(X, w, b):
    Z = np.dot(X, w) + b
    A = sigmoid(Z)
    return A

def compute_cost(A, Y):
    lambda_reg = 0.1
    m = len(X)
    loss = -np.sum(Y * np.log(A)+(1-Y)*np.log(1-A)) / m
    regularized_cost = lambda_reg * np.sum(np.square(w)) / (2 * m)

    cost = loss + regularized_cost
    return cost
def regulaized_gradient(X, Y, A, w, b):
    m = len(X)
    dZ = A - Y
    dw = (1/m) * np.dot(X.T, dZ) + (0.1/m) * w
    db = (1/m) * np.sum(dZ)
    return dw, db
def gradient_decent(X, Y, w, b, learning_rate=0.01, num_iterations=1000):
    for i in range(num_iterations):
        A = sigmoid_function(X, w, b)
        dw, db = regulaized_gradient(X, Y, A, w, b)
        w -= learning_rate * dw
        b -= learning_rate * db
    return w, b
X_norm = scaling(X)
A = sigmoid_function(X_norm, w, b)
cost = compute_cost(A , y)
print("X normalized:\n", X_norm)
print("Predictions:\n", A)
print("Cost:\n", cost)
print("Regularized Gradient:\n", regulaized_gradient(X_norm, y, A, w, b))
# Model ko train karein
w_trained, b_trained = gradient_decent(X_norm, y, w, b, learning_rate=0.1, num_iterations=1000)

print("\n--- Training Mukammal ---")
print("Trained Weights (w):", w_trained)
print("Trained Bias (b):", b_trained)

# Final predictions check karein
final_preds = sigmoid_function(X_norm, w_trained, b_trained)
print("Final Probabilities:\n", final_preds)
print("Classes (0 ya 1):\n", (final_preds >= 0.5).astype(int))
```
